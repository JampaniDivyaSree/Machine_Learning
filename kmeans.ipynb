```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
iris=datasets.load_iris()
x=iris.data
```

```python
silhouette_scores=[]
inertia_values=[]
r=[2,3,4,5,6,7,8,9,10]
for cluster in r:
  kmeans=KMeans(n_clusters=cluster, n_init=10,random_state=42)
  kmeans.fit(x)
  cluster_labels=kmeans.labels_
  silhouette_avg=silhouette_score(x,cluster_labels)
  silhouette_scores.append(silhouette_avg)
  inertia=kmeans.inertia_
  inertia_values.append(inertia)
  print("For",cluster,"clusters","The average silhouette_score is:",silhouette_avg)
```

```python
plt.plot(r,silhouette_scores,marker='o')
plt.title('Silhouette scores for different number of clusters')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette score')
plt.show()
```

```python
plt.plot(r, inertia_values, marker='o')
plt.title('Within-Cluster Sum of Squares (WCSS) for different number of clusters')
plt.xlabel('Number of clusters (K)')
plt.ylabel('WCSS')

plt.show()
```

```python
print("Inertia values")
print()
for i in r:
  print("cluster{}: {}".format(i,inertia_values[i-2]))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score

# Load the Iris dataset
iris = datasets.load_iris()
X = iris.data

# List to store silhouette scores
silhouette_scores = []

# Range of components to try
components_range = list(range(2, 11))

for n_components in components_range:
    # Create a GaussianMixture instance with n_components
    gmm = GaussianMixture(n_components=n_components, random_state=42)
    # Fit the model to the data
    gmm.fit(X)
    # Get cluster labels
    cluster_labels = gmm.predict(X)
    # Calculate silhouette score
    silhouette_avg = silhouette_score(X, cluster_labels)
    silhouette_scores.append(silhouette_avg)
    print("For", n_components, "components The average silhouette_score is:", silhouette_avg)

# Plotting silhouette scores
plt.plot(components_range, silhouette_scores, marker='o')
plt.title('Silhouette scores for different number of components')
plt.xlabel('Number of components')
plt.ylabel('Silhouette score')
plt.show()
```
